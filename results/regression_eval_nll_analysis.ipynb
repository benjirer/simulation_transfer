```python
import wandb
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-colorblind")

# Initialize wandb API
api = wandb.Api()

# List of projects to include
projects = ["shape_tracing_new_data_v24"]

# Prepare data storage
data = []

for project in projects:
    runs = api.runs(f'{project}')
    for run in runs:
        # Get config variables
        config = run.config
        use_sim_model = config.get('use_sim_model')
        use_sim_prior = config.get('use_sim_prior')
        predict_difference = config.get('predict_difference')
        random_seed = config.get('random_seed')
        num_offline_collected_transitions = config.get('num_offline_collected_transitions')
        run_id = run.id

        # Ensure all required config variables are present
        if None in [use_sim_model, use_sim_prior, predict_difference, random_seed, num_offline_collected_transitions]:
            print(f'Warning: Missing config for run: {run_id}')
            continue  # Skip runs with missing config

        # Get the history (metrics logged during the run)
        history = run.history(keys=['_step', 'eval/episode_reward'], pandas=True)

        # Check if history is empty
        if history.empty:
            print(f'Warning: No history for run: {run_id}')
            continue

        # Get the value at the maximum 'x_axis/bnn_step'
        max_step = history['_step'].max()
        last_reward = history.loc[history['_step'] == max_step, 'eval/episode_reward'].values

        # Get best reward if no value at max step
        if len(last_reward) == 0:
            last_reward = [history['eval/episode_reward'].max()]

        if len(last_reward) == 0:
            print(f'Warning: No eval/episode_reward available for run: {run_id}')
            continue  # Skip if no 'eval/episode_reward' found

        last_reward = last_reward[0]

        # Append to data
        data.append({
            'use_sim_model': use_sim_model,
            'use_sim_prior': use_sim_prior,
            'predict_difference': predict_difference,
            'random_seed': random_seed,
            'num_offline_collected_transitions': num_offline_collected_transitions,
            'eval_reward': last_reward
        })

# Convert data to DataFrame
df = pd.DataFrame(data)

# Map the models according to your specifications
def classify_model(row):
    if row['use_sim_model'] == 1 and row['use_sim_prior'] == 0:
        return 'SIM-MODEL'
    elif row['use_sim_model'] == 0 and row['use_sim_prior'] == 1:
        return 'BNN-SIM-FSVGD'
    elif row['use_sim_model'] == 0 and row['use_sim_prior'] == 0:
        return 'BNN-FSVGD'
    else:
        print(f'Warning: Unknown configuration for row: {row}')
        return None  # Exclude other configurations

df['Model'] = df.apply(classify_model, axis=1)

# Filter out runs that do not match any of the specified models
df = df[df['Model'].notnull()]

# Apply filter based on predict_difference
conditions = (
    ((df['Model'] == 'BNN-FSVGD') & (df['predict_difference'] == 1)) |
    ((df['Model'] == 'BNN-SIM-FSVGD') & (df['predict_difference'] == 1)) |
    ((df['Model'] == 'SIM-MODEL') & (df['predict_difference'] == 1))
)
df = df[conditions]

# Define the sample sizes and seeds of interest
sample_sizes = [500, 1000, 2000, 3000, 4000, 5000]
seeds = [137, 332, 417]

# Filter data for the specified sample sizes and seeds
df = df[df['num_offline_collected_transitions'].isin(sample_sizes)]
df = df[df['random_seed'].isin(seeds)]

# Group data by Model and Sample Size
grouped = df.groupby(['Model', 'num_offline_collected_transitions'])

# Calculate mean, std, and standard error of 'eval_reward' for each group
stats = grouped['eval_reward'].agg(['mean', 'std', 'count']).reset_index()
stats['sem'] = stats['std'] / np.sqrt(stats['count'])

# Plotting
plt.figure(figsize=(10, 6))
models = ['BNN-SIM-FSVGD', 'SIM-MODEL', 'BNN-FSVGD']
colors = plt.get_cmap('tab10').colors
color_dict = dict(zip(models, colors))

for model in models:
    model_data = stats[stats['Model'] == model]
    # Sort the data by sample size for proper plotting
    model_data = model_data.sort_values('num_offline_collected_transitions')

    # Plot mean line
    plt.plot(
        model_data['num_offline_collected_transitions'],
        model_data['mean'],
        label=model,
        color=color_dict[model],
        marker='o'
    )

    # Fill between mean ± SEM
    plt.fill_between(
        model_data['num_offline_collected_transitions'],
        model_data['mean'] - model_data['sem'],
        model_data['mean'] + model_data['sem'],
        color=color_dict[model],
        alpha=0.2
    )

plt.xlabel('Number of Datapoints')
plt.ylabel('Episode Reward')
plt.title('Episode Reward over Sample Size\n(Average across all seeds per model)')
plt.xticks(sample_sizes)
plt.legend()
plt.grid(True)
current_size = plt.gcf().get_size_inches()
plt.gcf().set_size_inches(current_size[0]*1.3, current_size[1])
plt.tight_layout()
plt.show()
```
