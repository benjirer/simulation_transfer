```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set Seaborn style
plt.style.use("seaborn-v0_8-colorblind")

"""Collected Data"""
file_names_ellipse = {
    'bnn-sim-fsvgd': {
        1000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_1000_1_ellipse_large_sparse_ir1k7sxx_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_1000_2_ellipse_large_sparse_rmjkshzy_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_1000_3_ellipse_large_sparse_hlafbsjp_v/brax_transitions.pickle",
        },
        4000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_4000_1_ellipse_large_sparse_n7vsto4d_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_4000_2_ellipse_large_sparse_usd0rgvm_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_4000_3_ellipse_large_sparse_u1e4pwkz_v/brax_transitions.pickle",
        },
        8000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_8000_1_ellipse_large_sparse_jrnpnh2e_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_8000_2_ellipse_large_sparse_n56lp7fu_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_8000_3_ellipse_large_sparse_qmvn3cn2_v/brax_transitions.pickle",
        },
    },
    'sim-model': {
        1000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_1000_1_ellipse_large_sparse_b1y879ru_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_1000_2_ellipse_large_sparse_gaej86dr_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_1000_3_ellipse_large_sparse_4pg8teow_v/brax_transitions.pickle",
        },
        4000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_4000_1_ellipse_large_sparse_nozm88or_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_4000_2_ellipse_large_sparse_xk7rephs_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_4000_3_ellipse_large_sparse_rvg6ncag_v/brax_transitions.pickle", 
        },
        8000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_8000_1_ellipse_large_sparse_vbvslmcc_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_8000_2_ellipse_large_sparse_ijqk8mfs_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_8000_3_ellipse_large_sparse_tflc8k9d_v/brax_transitions.pickle",
        },
    },
    'bnn-fsvgd': {
        1000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_1000_1_ellipse_large_sparse_84hxv3eb_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_1000_2_ellipse_large_sparse_zkn8ocns_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_1000_3_ellipse_large_sparse_qpltcy4p_v/brax_transitions.pickle",
        },
        4000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_4000_1_ellipse_large_sparse_yqgtomfy_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_4000_2_ellipse_large_sparse_mqv3ftob_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_4000_3_ellipse_large_sparse_11xyv7vl_v/brax_transitions.pickle",
        },
        8000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_8000_1_ellipse_large_sparse_q0naoo0u_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_8000_2_ellipse_large_sparse_ur92ckf7_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_8000_3_ellipse_large_sparse_96djmm04_v/brax_transitions.pickle",
        },
    },
}

"""Goal Trajectory"""
gt_file_name_ellipse = "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/goal_traj/ellipse_large_sparse_goal_trajectory.pkl"
gt_file_name_infinity = "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/goal_traj/infinity_goal_trajectory.pkl"
gt_file_name_heart = "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/goal_traj/heart_goal_trajectory.pkl"

shape_option = 'ellipse'

if shape_option == 'ellipse':
    file_names = file_names_ellipse
    gt_file_name = gt_file_name_ellipse
elif shape_option == 'infinity':
    file_names = file_names_infinity
    gt_file_name = gt_file_name_infinity
elif shape_option == 'heart':
    file_names = file_names_heart
    gt_file_name = gt_file_name_heart
else:
    raise ValueError(f"Invalid shape option: {shape_option}")

# Load the goal trajectory
with open(gt_file_name, "rb") as f:
    gt_data = pickle.load(f)

# Extract goal trajectory components
gt_ee_x = gt_data[..., 0]
gt_ee_y = gt_data[..., 1]
gt_ee_z = gt_data[..., 2]

# Initialize results dictionary
results = {}

# Process each file
for model_name in file_names:
    results[model_name] = {}
    for data_amount in file_names[model_name]:
        results[model_name][data_amount] = {}
        for seed_id in file_names[model_name][data_amount]:
            file_name = file_names[model_name][data_amount][seed_id]
            try:
                # Load the data
                with open(file_name, "rb") as f:
                    data = pickle.load(f)
                observations = np.array([d.observation for d in data])

                # Truncate observations and gt_data to same length
                length = min(observations.shape[0], gt_data.shape[0])
                observations = observations[:length]
                gt_data_trunc = gt_data[:length]

                # Extract x, y, z
                ee_x = observations[:, 7]
                ee_y = observations[:, 8]
                ee_z = observations[:, 9]
                gt_ee_x_trunc = gt_data_trunc[..., 0]
                gt_ee_y_trunc = gt_data_trunc[..., 1]
                gt_ee_z_trunc = gt_data_trunc[..., 2]

                # Calculate error
                error = np.linalg.norm(
                    np.array([ee_x, ee_y, ee_z]) - np.array([gt_ee_x_trunc, gt_ee_y_trunc, gt_ee_z_trunc]), axis=0
                )
                # Total error over the trajectory
                total_error = np.mean(error)

                # Store the data
                results[model_name][data_amount][seed_id] = {
                    'ee_y': ee_y,
                    'ee_z': ee_z,
                    'error': error,
                    'total_error': total_error,
                }
            except FileNotFoundError:
                print(f"File not found: {file_name}")
                continue

# Define color palette
models = list(results.keys())
colors = sns.color_palette("tab10", n_colors=len(models))
model_colors = dict(zip(models, colors))

# Prepare data for y-z trajectory plot
trajectory_data = {}
for model_name in results:
    trajectory_data[model_name] = {}
    for data_amount in results[model_name]:

        # Collect all ee_y and ee_z from different seeds
        ee_ys = []
        ee_zs = []
        min_length = None
        for seed_id in results[model_name][data_amount]:
            ee_y = results[model_name][data_amount][seed_id]['ee_y']
            ee_z = results[model_name][data_amount][seed_id]['ee_z']
            if min_length is None or len(ee_y) < min_length:
                min_length = len(ee_y)
            ee_ys.append(ee_y)
            ee_zs.append(ee_z)

        # Truncate all to min_length
        ee_ys = [ee_y[:min_length] for ee_y in ee_ys]
        ee_zs = [ee_z[:min_length] for ee_z in ee_zs]

        # Compute mean and std
        ee_y_mean = np.mean(ee_ys, axis=0)
        ee_y_std = np.std(ee_ys, axis=0)
        ee_z_mean = np.mean(ee_zs, axis=0)
        ee_z_std = np.std(ee_zs, axis=0)
        trajectory_data[model_name][data_amount] = {
            'ee_y_mean': ee_y_mean,
            'ee_y_std': ee_y_std,
            'ee_z_mean': ee_z_mean,
            'ee_z_std': ee_z_std,
        }

# Prepare data for error over time plot
error_time_data = {}
for model_name in results:
    error_time_data[model_name] = {}
    for data_amount in results[model_name]:
        # Collect all error arrays from different seeds
        errors = []
        min_length = None
        for seed_id in results[model_name][data_amount]:
            error = results[model_name][data_amount][seed_id]['error']
            if min_length is None or len(error) < min_length:
                min_length = len(error)
            errors.append(error)
        # Truncate all to min_length
        errors = [error[:min_length] for error in errors]
        # Compute mean and std
        error_mean = np.mean(errors, axis=0)
        error_std = np.std(errors, axis=0)
        error_time_data[model_name][data_amount] = {
            'error_mean': error_mean,
            'error_std': error_std,
        }

# Map data amounts to line styles
data_sizes = sorted(list(next(iter(results.values())).keys()))
data_amount_styles = {1000: ':', 4000: '--', 8000: '-'}

# Plot y-z trajectory
plt.figure(figsize=(10, 6))
# Plot the goal trajectory
plt.plot(gt_ee_y, gt_ee_z, label="Target", linestyle='--', color='red')

# Plot the trajectories from the results
for model_name in models:
    for data_amount in data_sizes:
        data = trajectory_data[model_name].get(data_amount, None)
        if data:
            ee_y_mean = data['ee_y_mean']
            ee_z_mean = data['ee_z_mean']
            label = f"{model_name}, data={data_amount}"
            color = model_colors[model_name]
            line_style = data_amount_styles[data_amount]
            # Plot mean trajectory
            plt.plot(ee_y_mean, ee_z_mean, label=label, color=color, linestyle=line_style)
plt.title("Y-Z Trajectory")
plt.xlabel("Y")
plt.ylabel("Z")
# Place legend in center right
plt.legend(loc='center right')
plt.show()

# Plot error over time
plt.figure(figsize=(10, 6))

for model_name in models:
    for data_amount in data_sizes:
        data = error_time_data[model_name].get(data_amount, None)
        if data:
            error_mean = data['error_mean']
            error_std = data['error_std']
            label = f"{model_name}, data={data_amount}"
            color = model_colors[model_name]
            line_style = data_amount_styles[data_amount]
            # Generate time steps
            timesteps = np.arange(len(error_mean))
            # Plot mean error
            plt.plot(timesteps, error_mean, label=label, color=color, linestyle=line_style)

plt.title("Error to Target Over Time")
plt.xlabel("Timesteps")
plt.ylabel("Error")
# Place legend in center right
plt.legend(loc='center right')
plt.show()

# Prepare data for total error plot
total_errors = {model: [] for model in models}
error_stds = {model: [] for model in models}

for model_name in models:
    for data_amount in data_sizes:
        errors = []
        for seed_id in results[model_name][data_amount]:
            errors.append(results[model_name][data_amount][seed_id]['total_error'])
        if errors:
            mean_error = np.mean(errors)
            std_error = np.std(errors)
            total_errors[model_name].append(mean_error)
            error_stds[model_name].append(std_error)
        else:
            total_errors[model_name].append(np.nan)
            error_stds[model_name].append(np.nan)

# Plot total error vs data sample sizes
plt.figure(figsize=(10, 6))

for model_name in models:
    x = data_sizes
    y = total_errors[model_name]
    y_lower = [mean - std for mean, std in zip(y, error_stds[model_name])]
    y_upper = [mean + std for mean, std in zip(y, error_stds[model_name])]
    plt.plot(
        x,
        y,
        marker='o',
        label=model_name,
        color=model_colors[model_name]
    )
    plt.fill_between(
        x,
        y_lower,
        y_upper,
        color=model_colors[model_name],
        alpha=0.2
    )

plt.xticks(data_sizes)
plt.title("Total Error vs Data Sample Sizes")
plt.xlabel("Data Sample Size")
plt.ylabel("Mean Total Error (across seeds)")
plt.legend()
plt.show()
```

```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from mpl_toolkits.mplot3d import Axes3D  # For 3D plotting

# Set Seaborn style
plt.style.use("seaborn-v0_8-colorblind")

"""Collected Data"""
file_names_ellipse = {
    'bnn-sim-fsvgd': {
        1000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_1000_1_ellipse_large_sparse_ir1k7sxx_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_1000_2_ellipse_large_sparse_rmjkshzy_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_1000_3_ellipse_large_sparse_hlafbsjp_v/brax_transitions.pickle",
        },
        4000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_4000_1_ellipse_large_sparse_n7vsto4d_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_4000_2_ellipse_large_sparse_usd0rgvm_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_4000_3_ellipse_large_sparse_u1e4pwkz_v/brax_transitions.pickle",
        },
        8000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_8000_1_ellipse_large_sparse_jrnpnh2e_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_8000_2_ellipse_large_sparse_n56lp7fu_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-sim-fsvgd_8000_3_ellipse_large_sparse_qmvn3cn2_v/brax_transitions.pickle",
        },
    },
    'sim-model': {
        1000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_1000_1_ellipse_large_sparse_b1y879ru_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_1000_2_ellipse_large_sparse_gaej86dr_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_1000_3_ellipse_large_sparse_4pg8teow_v/brax_transitions.pickle",
        },
        4000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_4000_1_ellipse_large_sparse_nozm88or_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_4000_2_ellipse_large_sparse_xk7rephs_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_4000_3_ellipse_large_sparse_rvg6ncag_v/brax_transitions.pickle", 
        },
        8000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_8000_1_ellipse_large_sparse_vbvslmcc_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_8000_2_ellipse_large_sparse_ijqk8mfs_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_sim-model_8000_3_ellipse_large_sparse_tflc8k9d_v/brax_transitions.pickle",
        },
    },
    'bnn-fsvgd': {
        1000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_1000_1_ellipse_large_sparse_84hxv3eb_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_1000_2_ellipse_large_sparse_zkn8ocns_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_1000_3_ellipse_large_sparse_qpltcy4p_v/brax_transitions.pickle",
        },
        4000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_4000_1_ellipse_large_sparse_yqgtomfy_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_4000_2_ellipse_large_sparse_mqv3ftob_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_4000_3_ellipse_large_sparse_11xyv7vl_v/brax_transitions.pickle",
        },
        8000: {
            1: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_8000_1_ellipse_large_sparse_q0naoo0u_v/brax_transitions.pickle",
            2: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_8000_2_ellipse_large_sparse_ur92ckf7_v/brax_transitions.pickle",
            3: "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/collected_data/shape_tracing_v9/ellipse_large_sparse/test_shape_tracing_v9_bnn-fsvgd_8000_3_ellipse_large_sparse_96djmm04_v/brax_transitions.pickle",
        },
    },
}

"""Goal Trajectory"""
gt_file_name_ellipse = "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/goal_traj/ellipse_large_sparse_goal_trajectory.pkl"
gt_file_name_infinity = "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/goal_traj/infinity_goal_trajectory.pkl"
gt_file_name_heart = "/home/bhoffman/Documents/MT FS24/active-learning-dynamics/goal_traj/heart_goal_trajectory.pkl"

shape_option = 'ellipse'

if shape_option == 'ellipse':
    file_names = file_names_ellipse
    gt_file_name = gt_file_name_ellipse
else:
    raise ValueError(f"Invalid shape option: {shape_option}")

# Load the goal trajectory
with open(gt_file_name, "rb") as f:
    gt_data = pickle.load(f)

# Extract goal trajectory components
gt_ee_x = gt_data[..., 0]
gt_ee_y = gt_data[..., 1]
gt_ee_z = gt_data[..., 2]

# Initialize results dictionary
results = {}

# Process each file
for model_name in file_names:
    results[model_name] = {}
    for data_amount in file_names[model_name]:
        results[model_name][data_amount] = {}
        for seed_id in file_names[model_name][data_amount]:
            file_name = file_names[model_name][data_amount][seed_id]
            try:
                # Load the data
                with open(file_name, "rb") as f:
                    data = pickle.load(f)
                observations = np.array([d.observation for d in data])

                obs_length = observations.shape[0]
                gt_length = gt_data.shape[0]

                if obs_length < gt_length:
                    print(f"Warning: Observations shorter than goal trajectory for {file_name}. Repeating last error.")

                # Extract x, y, z
                ee_x = observations[:, 7]
                ee_y = observations[:, 8]
                ee_z = observations[:, 9]

                # Truncate or extend ee_x, ee_y, ee_z to match gt_length
                if obs_length >= gt_length:
                    # Truncate observations to gt_length
                    ee_x = ee_x[:gt_length]
                    ee_y = ee_y[:gt_length]
                    ee_z = ee_z[:gt_length]
                else:
                    # Extend observations to gt_length by repeating last value
                    last_x = ee_x[-1]
                    last_y = ee_y[-1]
                    last_z = ee_z[-1]
                    ee_x = np.concatenate([ee_x, np.full(gt_length - obs_length, last_x)])
                    ee_y = np.concatenate([ee_y, np.full(gt_length - obs_length, last_y)])
                    ee_z = np.concatenate([ee_z, np.full(gt_length - obs_length, last_z)])

                # Extract gt data
                gt_ee_x_trunc = gt_data[..., 0]
                gt_ee_y_trunc = gt_data[..., 1]
                gt_ee_z_trunc = gt_data[..., 2]

                # Calculate error
                error = np.linalg.norm(
                    np.array([ee_x, ee_y, ee_z]) - np.array([gt_ee_x_trunc, gt_ee_y_trunc, gt_ee_z_trunc]), axis=0
                )
                # Total error over the trajectory
                total_error = np.mean(error)

                # Store the data
                results[model_name][data_amount][seed_id] = {
                    'ee_x': ee_x,
                    'ee_y': ee_y,
                    'ee_z': ee_z,
                    'error': error,
                    'total_error': total_error,
                }
            except FileNotFoundError:
                print(f"File not found: {file_name}")
                continue

# Define color palette
models = list(results.keys())
colors = sns.color_palette("tab10", n_colors=len(models))
model_colors = dict(zip(models, colors))

# Map data amounts to line styles
data_sizes = sorted(list(next(iter(results.values())).keys()))
data_amount_styles = {1000: ':', 4000: '--', 8000: '-'}

# Prepare data for trajectory plots
trajectory_data = {}
for model_name in results:
    trajectory_data[model_name] = {}
    for data_amount in results[model_name]:
        trajectory_data[model_name][data_amount] = {}
        for seed_id in results[model_name][data_amount]:
            data = results[model_name][data_amount][seed_id]
            ee_x = data['ee_x']
            ee_y = data['ee_y']
            ee_z = data['ee_z']
            trajectory_data[model_name][data_amount][seed_id] = {
                'ee_x': ee_x,
                'ee_y': ee_y,
                'ee_z': ee_z,
            }

# Prepare data for error over time plot
error_time_data = {}
for model_name in results:
    error_time_data[model_name] = {}
    for data_amount in results[model_name]:
        error_time_data[model_name][data_amount] = {}
        for seed_id in results[model_name][data_amount]:
            error = results[model_name][data_amount][seed_id]['error']
            error_time_data[model_name][data_amount][seed_id] = error

# Plot Y-Z Trajectory
plt.figure(figsize=(10, 6))
# Plot the goal trajectory
plt.plot(gt_ee_y, gt_ee_z, label="Target", linestyle='--', color='red')

labeled = set()

# Plot the trajectories from the results
for model_name in models:
    for data_amount in data_sizes:
        data_seeds = trajectory_data[model_name].get(data_amount, None)
        if data_seeds:
            for seed_id, seed_data in data_seeds.items():
                ee_y = seed_data['ee_y']
                ee_z = seed_data['ee_z']
                color = model_colors[model_name]
                line_style = data_amount_styles[data_amount]
                key = (model_name, data_amount)
                if key not in labeled:
                    label = f"{model_name}, data={data_amount}"
                    labeled.add(key)
                else:
                    label = None
                # Plot trajectory for this seed
                plt.plot(ee_y, ee_z, color=color, linestyle=line_style, alpha=0.5, label=label)
plt.title("Y-Z Trajectory")
plt.xlabel("Y")
plt.ylabel("Z")
plt.legend(loc='center right')
plt.show()

# Plot X-Y Trajectory
plt.figure(figsize=(10, 6))
# Plot the goal trajectory
plt.plot(gt_ee_x, gt_ee_y, label="Target", linestyle='--', color='red')

labeled = set()

for model_name in models:
    for data_amount in data_sizes:
        data_seeds = trajectory_data[model_name].get(data_amount, None)
        if data_seeds:
            for seed_id, seed_data in data_seeds.items():
                ee_x = seed_data['ee_x']
                ee_y = seed_data['ee_y']
                color = model_colors[model_name]
                line_style = data_amount_styles[data_amount]
                key = (model_name, data_amount)
                if key not in labeled:
                    label = f"{model_name}, data={data_amount}"
                    labeled.add(key)
                else:
                    label = None
                plt.plot(ee_x, ee_y, color=color, linestyle=line_style, alpha=0.5, label=label)
plt.title("X-Y Trajectory")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend(loc='center right')
plt.show()

# Plot X-Z Trajectory
plt.figure(figsize=(10, 6))
# Plot the goal trajectory
plt.plot(gt_ee_x, gt_ee_z, label="Target", linestyle='--', color='red')

labeled = set()

for model_name in models:
    for data_amount in data_sizes:
        data_seeds = trajectory_data[model_name].get(data_amount, None)
        if data_seeds:
            for seed_id, seed_data in data_seeds.items():
                ee_x = seed_data['ee_x']
                ee_z = seed_data['ee_z']
                color = model_colors[model_name]
                line_style = data_amount_styles[data_amount]
                key = (model_name, data_amount)
                if key not in labeled:
                    label = f"{model_name}, data={data_amount}"
                    labeled.add(key)
                else:
                    label = None
                plt.plot(ee_x, ee_z, color=color, linestyle=line_style, alpha=0.5, label=label)
plt.title("X-Z Trajectory")
plt.xlabel("X")
plt.ylabel("Z")
plt.legend(loc='center right')
plt.show()

# Plot 3D Trajectory
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

# Plot the goal trajectory
ax.plot(gt_ee_x, gt_ee_y, gt_ee_z, label="Target", linestyle='--', color='red')

labeled = set()

for model_name in models:
    for data_amount in data_sizes:
        data_seeds = trajectory_data[model_name].get(data_amount, None)
        if data_seeds:
            for seed_id, seed_data in data_seeds.items():
                ee_x = seed_data['ee_x']
                ee_y = seed_data['ee_y']
                ee_z = seed_data['ee_z']
                color = model_colors[model_name]
                line_style = data_amount_styles[data_amount]
                key = (model_name, data_amount)
                if key not in labeled:
                    label = f"{model_name}, data={data_amount}"
                    labeled.add(key)
                else:
                    label = None
                ax.plot(ee_x, ee_y, ee_z, color=color, linestyle=line_style, alpha=0.5, label=label)
# Add labels, title, legend
ax.set_title("3D Trajectory")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.legend(loc='upper left')
plt.show()

# Plot error over time
plt.figure(figsize=(10, 6))

labeled = set()

for model_name in models:
    for data_amount in data_sizes:
        data_seeds = error_time_data[model_name].get(data_amount, None)
        if data_seeds:
            for seed_id, error in data_seeds.items():
                color = model_colors[model_name]
                line_style = data_amount_styles[data_amount]
                # Generate time steps
                timesteps = np.arange(len(error))
                # Determine label
                key = (model_name, data_amount)
                if key not in labeled:
                    label = f"{model_name}, data={data_amount}"
                    labeled.add(key)
                else:
                    label = None
                # Plot error for this seed
                plt.plot(timesteps, error, color=color, linestyle=line_style, alpha=0.5, label=label)
plt.title("Error to Target Over Time")
plt.xlabel("Timesteps")
plt.ylabel("Error")
plt.legend(loc='center right')
plt.show()

# Prepare data for total error plot
total_errors = {model: [] for model in models}
error_stds = {model: [] for model in models}

for model_name in models:
    for data_amount in data_sizes:
        errors = []
        for seed_id in results[model_name][data_amount]:
            errors.append(results[model_name][data_amount][seed_id]['total_error'])
        if errors:
            mean_error = np.mean(errors)
            std_error = np.std(errors)
            total_errors[model_name].append(mean_error)
            error_stds[model_name].append(std_error)
        else:
            total_errors[model_name].append(np.nan)
            error_stds[model_name].append(np.nan)

# Plot total error vs data sample sizes
plt.figure(figsize=(10, 6))

for model_name in models:
    x = data_sizes
    y = total_errors[model_name]
    y_lower = [mean - std for mean, std in zip(y, error_stds[model_name])]
    y_upper = [mean + std for mean, std in zip(y, error_stds[model_name])]
    plt.plot(
        x,
        y,
        marker='o',
        label=model_name,
        color=model_colors[model_name]
    )
    plt.fill_between(
        x,
        y_lower,
        y_upper,
        color=model_colors[model_name],
        alpha=0.2
    )

plt.xticks(data_sizes)
plt.title("Total Error vs Data Sample Sizes")
plt.xlabel("Data Sample Size")
plt.ylabel("Mean Total Error (across seeds)")
plt.legend()
plt.show()
```
